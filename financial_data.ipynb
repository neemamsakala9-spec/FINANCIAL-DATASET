```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('financial.csv')
df
```

```python
print(df.isnull().sum())
```

```python
numeric_columns = df.select_dtypes(include="number").columns
numeric_columns
```

```python
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())
    df[column]
```

#filling missing value in numerical columns using median


```python
df["zone"] = df["zone"].fillna(df["zone"].mode()[0])
df["zone"]
```

#filling missing categorical values using mode

```python
from sklearn.linear_model import LinearRegression

data = df[["banked_pct_2023", "formal_inclusion_pct_2023"]].dropna()

X = data[["banked_pct_2023"]]
y = data["formal_inclusion_pct_2023"]

model = LinearRegression()
model.fit(X, y)

prediction = model.predict([[30]])

print("Predicted formal inclusion:", prediction[0])
```
